```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection
from sklearn.metrics import r2_score
```

```python
X, y = datasets.load_diabetes(return_X_y=True)
print(X.shape)
print(X[0])
```

# Note
Values in data are normalised (around 0)
A NumPy array shape is (No. rows , No. cols)

## dataset
X → the input features (things the model uses)
y → the target (what the model predicts)

```python
X = X[:, 2]
X = X.reshape((-1,1))
```

```python
print(X.shape)
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

splitting intital dataset into training and test data (66-33) / hold-out validation

```python
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

Learning pattern between x and y / finding weights and etc

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test,  color='black')
plt.plot(X_test, y_pred, color='blue', linewidth=3)
plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title('A Graph Plot Showing Diabetes Progression Against BMI')
plt.show()
```
